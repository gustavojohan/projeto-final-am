```python
# ── 1. Remover NU_FASE_VIDA (constante) ───────────────────────────────────────
df = df.drop(columns=['NU_FASE_VIDA'])

# ── 2. Corrigir NU_PESO (vírgula → ponto) e converter ────────────────────────
df['NU_PESO']   = df['NU_PESO'].str.replace(',', '.', regex=False)
df['NU_PESO']   = pd.to_numeric(df['NU_PESO'],   errors='coerce')
df['NU_ALTURA'] = pd.to_numeric(df['NU_ALTURA'], errors='coerce')

antes = len(df)
df = df.dropna(subset=['NU_PESO', 'NU_ALTURA'])
print(f"Removidos por peso/altura inválidos: {antes - len(df)}")

# ── 3. Calcular IMC ───────────────────────────────────────────────────────────
df['IMC'] = df['NU_PESO'] / (df['NU_ALTURA'] / 100) ** 2

# ── 4. Corrigir DT_ACOMPANHAMENTO (formato DD/MM/YYYY) ───────────────────────
df['DT_ACOMPANHAMENTO'] = pd.to_datetime(
    df['DT_ACOMPANHAMENTO'], format='%d/%m/%Y', errors='coerce'
)
df['ANO_ACOMP'] = df['DT_ACOMPANHAMENTO'].dt.year
df = df.drop(columns=['DT_ACOMPANHAMENTO'])

# ── 5. Criar feature de região ────────────────────────────────────────────────
df['regiao'] = df['NO_MUNICIPIO'].apply(mapear_regiao)

nao_mapeados = df[df['regiao'] == 'Não identificado']['NO_MUNICIPIO'].unique()
print(f"\nMunicípios não mapeados ({len(nao_mapeados)}):")
print(sorted(nao_mapeados))

df = df.drop(columns=['NO_MUNICIPIO'])

# ── 6. Encoding das categóricas ───────────────────────────────────────────────
df['SG_SEXO'] = df['SG_SEXO'].map({'M': 0, 'F': 1})

# Corrigido: 'E-SUS AB' (com hífen) conforme o CSV
df['SISTEMA_ORIGEM_ACOMP'] = df['SISTEMA_ORIGEM_ACOMP'].map({
    'SISVAN WEB':    0,
    'AUXILIO BRASIL': 1,
    'E-SUS AB':       2,
})

df['regiao'] = df['regiao'].map({
    'Sertão Paraibano':  0,
    'Borborema':         1,
    'Agreste Paraibano': 2,
    'Mata Paraibana':    3,
})

# ── 7. Encoding do target ─────────────────────────────────────────────────────
target_map = {
    'Baixo peso':            0,
    'Adequado ou eutrófico': 1,
    'Sobrepeso':             2,
    'Obesidade Grau I':      3,
    'Obesidade Grau II':     4,
    'Obesidade Grau III':    5,
}
df['target'] = df['CO_ESTADO_NUTRI_ADULTO'].map(target_map)
df = df.drop(columns=['CO_ESTADO_NUTRI_ADULTO'])

# ── 8. Resumo final ───────────────────────────────────────────────────────────
print(f"\nShape final: {df.shape}")
print(f"\nNulos restantes:\n{df.isna().sum()}")
print(f"\nDistribuição do target:\n{df['target'].value_counts().sort_index()}")
```
